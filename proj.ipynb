**Personalized AI Travel Planner**

```python
from langchain_openai import ChatOpenAI
from sec import openapi_key

import os
os.environ["Private"] = openapi_key

llm = ChatOpenAI(
    model="nvidia/nemotron-3-super-120b-a12b:free",
    base_url="https://openrouter.ai/api/v1",
    api_key=openapi_key,
    temperature=0.6
)

# just api woek checker
app = llm.invoke("Hi just give me output 1-10 nothing else")
```

```python
print(app.content)
```

```python
from langchain_core.prompts import PromptTemplate

first = PromptTemplate(
    input_variables=['Travelpreferences'],
    template="Based on my travel preferences: **{Travelpreferences}**, suggest a list of suitable travel destinations from around the world.For each destination, provideDestination NameCountryWhy Visit: A brief explanation of why it matches my preferences.Top Attractions: A few must-visit places or experiences.Finally, analyze all the suggested destinations and recommend one best destination that most closely matches my travel preferences. Explain why it is the best choice."
)

Travel = input("Enter your Travelpreferences: ")

first.format(Travelpreferences = Travel)
```

```python
chain = first | llm

Travel = input("Enter your Travelpreferences: ")

res = chain.invoke(Travel)
```

```python
from IPython.display import display, Markdown

display(Markdown(res.content))
```

```python
from langchain_classic.memory import ConversationBufferWindowMemory
from langchain_classic.chains import ConversationChain

memory = ConversationBufferWindowMemory(k = 5)

memo_agent = ConversationChain(
    llm = llm,
    memory = memory
)

chain2 = first | memo_agent

Travel = input("Enter your Travelpreferences: ")

response = memo_agent.predict(input=Travel)
display(Markdown(response))
```

```python
from pls import serp_api

os.environ["serch"] = serp_api

from langchain_community.utilities import SerpAPIWrapper

serch =  SerpAPIWrapper(
    serpapi_api_key= serp_api
)
from langchain_core.tools import Tool
serch_tool = Tool(
    name = "serch",
    description="Search the web for current information.",
    func=serch.run
)

from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()


agent = create_agent(
    model = llm,
    tools=[serch_tool],
    checkpointer=checkpointer,
    system_prompt="use this for research attractions, compare options, and refine the itinerary through conversation."
)

response = agent.invoke(
    {
        "messages": [
            {"role": "user", "content": "My favorite destination is Japan.They is attractions"}
        ]
    },
    config={"configurable": {"thread_id": "user1"}}
)

print(response["messages"][-1].content)


```
